## 5. Tuples and *unpacking*

A **tuple** is an immutable sequence typically written in parentheses.

**Unpacking** assigns the elements of a sequence to several variables at once. For example, use `x, y, v, theta = state` and `a, omega = u`.

````python
point = (3.0, 4.0, 6.0)
print(point[0], point[1])

# point[0] = 2.0  # This would cause an error since tuples are immutable

# Unpacking
px, py, pz, pw = point
print(f"px = {px}, py = {py}, pz = {pz}")

# Works with any sequence of the right length
state = [1.0, 2.0, 0.5, 0.0]   # [x, y, theta, v]
x, y, theta, v = state
print(x, y, theta, v)
````

> ⚠️ The number of variables on the left must match the number of elements in the sequence. **Order matters.** If the state is `[x, y, theta, v]` and you write `x, y, v, theta = state`, Python will assign the angle to `v` without raising an error.

### A first simulation with constant acceleration (double integrator)

Start at rest with constant acceleration and update velocity and position at each time step $\Delta t$.

$$v_{k+1} = v_k + a\,\Delta t \qquad p_{k+1} = p_k + v_k\,\Delta t$$

````python
dt = 0.5
a = 1.0
p, v = 0.0, 0.0   # multiple assignment

for k in range(6):
    t = k * dt
    print(f"t = {t:.1f} s   p = {p:.3f} m   v = {v:.2f} m/s")
    v += a * dt 
    p += v * dt
````

### Getting the index and element with `enumerate`

````python
measurements = [9.7, 9.9, 9.8]
for i, m in enumerate(measurements):
    print(f"measurement {i}: {m}")
````

### The `while` loop

A `while` loop repeats as long as a condition is true. Use it when you do not know in advance how many iterations you will need.

````python
# How long does it take to reach 10 m/s with a = 2 m/s^2?
v, t, dt = 0.0, 0.0, 0.1
while v < 10.0:
    v += 2.0 * dt
    t += dt
    print(f"t = {t:.1f} s   v = {v:.2f} m/s")
print(f"Reached {v:.2f} m/s after {t:.1f} s")
````

### Building a list inside a loop

Create an empty list **before** the loop and call `append` **inside** it.

````python
squares = []
for n in range(6):
    squares.append(n ** 2)
print(squares)

# Alternatively, using a list comprehension:
squares = [n ** 2 for n in range(6)]
print(squares)
for k in range(0, 11, 2):
    print(k, end=" ")
print()
````

Separate the values with commas in a `return` statement. Python returns them as a tuple, which you can unpack.

````python
import math

def polar_to_cartesian(r, angle):
    x = r * math.cos(angle)
    y = r * math.sin(angle)
    return x, y

x, y = polar_to_cartesian(2.0, 0.5)
print(x, y)
````

## 11. References and copies

In Python, a variable name refers to an object. With a list or NumPy array, `b = a` makes both names refer to the **same** object.

````python
a = [1, 2, 3]
b = a          # b and a refer to the SAME object
b[0] = 99
print("a =", a)   # a changed too
print(a is b)   # True
````

To get an independent object, make an explicit **copy**.

````python
a = [1, 2, 3]
b = a.copy()   # a new object
b[0] = 99
print("a =", a)
print("b =", b)
print(a is b)   # False
````

### Saving a copy of the simulation state

The simulation saves the state at each step with `x_log.append(x.copy())`. The example below shows what happens when you omit `.copy()`.

````python
state = [0.0]
history_without_copies = []
history_with_copies = []

for k in range(3):
    state[0] += 1.0                       # modify in place
    history_without_copies.append(state)        # store a reference to the same object
    history_with_copies.append(state.copy())    # store a snapshot of the current value

print("without copy:", history_without_copies)
print("with copy:  ", history_with_copies)
````

Without `.copy()`, every entry in the history refers to the **same** object and displays its final value.

## 12. NumPy vectors and matrices

Multiplying a Python list repeats its elements. For example, `[1, 2] * 2` gives `[1, 2, 1, 2]`. NumPy provides **arrays** for representing vectors and matrices and performing elementwise arithmetic.

````python
import numpy as np
values = [1.0, 2.0, 3.0]
vector = np.array([1.0, 2.0, 3.0])

print("values * 2   =", values * 2)
print("vector * 2 =", vector * 2)
print("vector + vector =", vector + vector)
print("values + values =", values + values)
print("vector ** 2 =", vector ** 2)
````

### Array shape for 1D vectors, column vectors, and matrices

An array's **shape** (`shape`) gives the number of elements along each dimension.

- `np.zeros(4)` has shape `(4,)` and creates a 1D vector with four elements.
- `np.zeros((4, 1))` has shape `(4, 1)` and creates a **column vector** with four rows and one column.
- `np.zeros((2, 3))` creates a 2 × 3 matrix.

In `np.zeros((4, 1))`, the inner parentheses form the tuple passed as the shape.

````python
v1 = np.zeros(4)
v_col = np.zeros((4, 1))
M = np.array([[1, 2, 3],
              [4, 5, 6]])

print("v1.shape    =", v1.shape)
print("v_col.shape =", v_col.shape)
print("M.shape     =", M.shape)
print(v_col)
````

### 2D indexing with `A[row, column]`

For a matrix, separate the row and column indices with a comma. Indices start at 0. In a column vector of shape `(4, 1)`, the only column has index 0, so element $i$ is `x[i, 0]`.

For example, `x[3, 0]` gives the velocity $v$ and `x[2, 0]` gives the angle $\theta$ in the state vector $\mathbf{x} = [x, y, \theta, v]^T$.

````python
M = np.array([[1, 2, 3], [4, 5, 6]])
print(M[0, 2])   # row 0, column 2 -> 3
print(M[1, 0])   # row 1, column 0 -> 4

state = np.array([[1.0], [2.0], [0.3], [1.5]])   # column vector [x, y, theta, v]
print("theta =", state[2, 0])
print("v     =", state[3, 0])
````

### 2D slicing with `:`

A colon on its own selects every element along that dimension.

- `A[:, 0]` selects all of column 0.
- `A[1, :]` selects all of row 1.
- `A[-1, 0]` selects column 0 of the last row.

Use this syntax to extract all $x$ values (`x_log[:, 0]`) and $y$ values (`x_log[:, 1]`) from the simulation history.

````python
# Each row is a time step and each column is a variable [x, y]
trajectory = np.array([[0.0, 0.0],
                        [1.0, 0.5],
                        [2.0, 1.8],
                        [3.0, 4.1]])

print("all x values:     ", trajectory[:, 0])
print("all y values:     ", trajectory[:, 1])
print("first point:    ", trajectory[0, :])
print("last point:   ", trajectory[-1, :])
print("x of last point:  ", trajectory[-1, 0])
````

### Converting a list of arrays to a 2D array

During the simulation, save the states in a **list**. Afterward, convert the list to an array with `np.array` and reshape it to 2D so you can slice it with `[:, 0]`.

````python
log = []
s = np.zeros((4, 1))
for k in range(3):
    s += 1.0
    log.append(s.copy())

log = np.array(log)
print("shape after np.array:", log.shape)
log = log.squeeze()
print("shape after squeeze: ", log.shape)
print(log)

print("flattened s:", s.flatten())
````

## 14. Putting it all together

### 14.1 Forward Euler integration

A differential equation $\dot{\mathbf{x}} = f(\mathbf{x}, \mathbf{u})$ describes a dynamical system. To simulate it, advance time in discrete steps of length $\Delta t$ and approximate the next state as follows.

$$\mathbf{x}_{k+1} = \mathbf{x}_k + f(\mathbf{x}_k, \mathbf{u}_k)\,\Delta t$$

**Example** — A cart on a track has state $\mathbf{x} = [p,\ v]^T$ (position and velocity) and input $u = a$ (acceleration). Its dynamics are

$$\dot{p} = v \qquad \dot{v} = a$$

1. Set the parameters and initial state.
2. Define `f(x, u)` to return the state derivative.
3. Use a `for` loop to update the state with `x += f(x, u) * dt` and save a snapshot with `.copy()`.
4. Convert the log to an array and plot it.

````python
# 1. Parameters and initial state
dt = 0.1                          # integration time step [s]
x = np.array([0.0, 0.0])          # initial state [p, v]
u = 0.5                           # constant acceleration [m/s^2]

# 2. Dynamics
def f(x, u):
    p, v = x                      # unpack the state
    dp = v
    dv = u
    return np.array([dp, dv])

# 3. Simulation
x_log = []
for k in range(100):
    x += f(x, u) * dt
    x_log.append(x.copy())

# 4. Plot
x_log = np.array(x_log)
times = np.arange(1, 101) * dt

plt.figure()
plt.plot(times, x_log[:, 0], label="position p (m)")
plt.plot(times, x_log[:, 1], label="velocity v (m/s)")
plt.xlabel("Time (s)")
plt.legend()
plt.grid()
plt.show()

print("Final state:", x)
print("Exact solution p(10) = 0.5*a*t^2 =", 0.5 * u * 10**2)
````

> 🔎 The difference between the Euler result and the analytical final position is the **discretization error**. Reduce `dt`, increase the number of steps proportionally to keep the duration fixed, and observe how the error changes.

### 14.2 A proportional controller for cruise control

We want the cart to reach a **target velocity** $v_{cmd}$. Compute the error and set the acceleration proportional to it.

$$e = v_{cmd} - v \qquad a = K_p\, e$$

At each step, compute the **control input**, then integrate the **dynamics**.

````python
dt = 0.1
K_p = 0.8          # proportional gain
v_cmd = 1.2        # target velocity [m/s]

x = np.array([0.0, 0.0], dtype=float)   # [p, v]

def dynamics(x, a):
    p, v = x
    return np.array([v, a], dtype=float)

def control(x, v_cmd, K_p):
    p, v = x
    e = v_cmd - v
    a = K_p * e
    return a

x_log = []
a_log = []
times = np.arange(0, 10, dt)
for k in range(len(times)):
    a = control(x, v_cmd, K_p)
    x += dynamics(x, a) * dt
    x_log.append(x.copy())
    a_log.append(a)

x_log = np.array(x_log)
a_log = np.array(a_log)

plt.figure()
plt.plot(times, x_log[:, 1], label="velocity (m/s)")
plt.plot(times, a_log, label="acceleration (m/s²)")
plt.axhline(v_cmd, ls="--", color="r", label="v_cmd")
plt.xlabel("Time (s)")
plt.ylabel("Velocity / Acceleration")
plt.title("Proportional controller")
plt.legend()
plt.grid()
plt.show()
````
