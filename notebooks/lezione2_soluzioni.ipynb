**Exercise 9.2.** Complete the function that returns **two** values from a list of measurements, the mean and the maximum. Hint — Use `sum()`, `len()`, and `max()`.

*Extra practice*

```python
def mean_and_max(measurements):
    mean = sum(measurements) / len(measurements)
    maximum = max(measurements)
    return mean, maximum

m, M = mean_and_max([9.7, 9.9, 9.8, 10.1])
print(f"mean = {m:.3f}, maximum = {M}")
```

**Exercise 9.3.** Write a **saturation** function that returns `u_max` when `u` exceeds `u_max`, returns `-u_max` when `u` is below `-u_max`, and otherwise returns `u`. Set the default value of `u_max` to `1.0`.

*Extra practice*

```python
def saturate(u, u_max=1.0):
    if u > u_max:
        return u_max
    elif u < -u_max:
        return -u_max
    else:
        return u

print(saturate(0.5))           # 0.5
print(saturate(3.0))           # 1.0
print(saturate(-7.0, 2.0))     # -2.0
```

## Section 14

**Exercise 14.1.** A cart starts at 2 m/s and slows down due to viscous friction, with dynamics $\dot{p} = v$ and $\dot{v} = -c\,v$. Complete the dynamics function, the Euler update, and the line that saves the state.

*Extra practice*

```python
dt = 0.1
c = 0.5                          # friction coefficient [1/s]
x = np.array([0.0, 2.0])         # [p, v]

def friction_dynamics(x, c):
    p, v = x
    dp = v
    dv = -c * v
    return np.array([dp, dv])

x_log = []
for k in range(100):
    x += friction_dynamics(x, c) * dt
    x_log.append(x.copy())

x_log = np.array(x_log)
plt.figure()
plt.plot(x_log[:, 0], label="position (m)")
plt.plot(x_log[:, 1], label="velocity (m/s)")
plt.xlabel("step k")
plt.legend()
plt.grid()
plt.show()
```

**Exercise 14.2.** Complete the proportional controller and the simulation loop. Compute the **control input** first, then integrate the **dynamics** at each step. This time the cart should **slow down** from 3 m/s to 1 m/s.

```python
dt = 0.1
K_p = 1.0
v_cmd = 1.0
x = np.array([0.0, 3.0])         # [p, v]

def dynamics(x, a):
    p, v = x
    return np.array([v, a])

def control(x, v_cmd, K_p):
    p, v = x
    e = v_cmd - v
    return K_p * e

v_log = []
for k in range(80):
    a = control(x, v_cmd, K_p)
    x += dynamics(x, a) * dt
    v_log.append(x[1])

plt.figure()
plt.plot(np.arange(80) * dt, v_log, label="v")
plt.axhline(v_cmd, ls="--", color="r", label="v_cmd")
plt.xlabel("Time (s)")
plt.legend()
plt.grid()
plt.show()
```
